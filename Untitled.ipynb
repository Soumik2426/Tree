```python
import os
from PIL import Image, UnidentifiedImageError
import pandas as pd
import hashlib

# Initialize lists
image_paths, widths, heights, corrupted = [], [], [], []

# Root dataset directory
root_dir = "Tree_Species_Dataset"

# Loop through images
for folder in os.listdir(root_dir):
    folder_path = os.path.join(root_dir, folder)
    if not os.path.isdir(folder_path):
        continue
    for file in os.listdir(folder_path):
        file_path = os.path.join(folder_path, file)
        try:
            with Image.open(file_path) as img:
                w, h = img.size
                widths.append(w)
                heights.append(h)
                image_paths.append(file_path)
        except UnidentifiedImageError:
            corrupted.append(file_path)

# Create DataFrame of valid images
df_sizes = pd.DataFrame({
    "path": image_paths,
    "width": widths,
    "height": heights
})

# Check for very small and very large images
small_imgs = df_sizes[(df_sizes['width'] < 150) | (df_sizes['height'] < 150)]
large_imgs = df_sizes[(df_sizes['width'] > 1000) | (df_sizes['height'] > 2000)]

# Check for duplicate images (by hash)
def get_hash(file_path):
    with open(file_path, 'rb') as f:
        return hashlib.md5(f.read()).hexdigest()

df_sizes['hash'] = df_sizes['path'].apply(get_hash)
duplicate_imgs = df_sizes[df_sizes.duplicated('hash', keep=False)]

# Results
print(f" Total images checked: {len(df_sizes)}")
print(f" Corrupted images found: {len(corrupted)}")
print(f" Duplicate images found: {len(duplicate_imgs)}")
print(f" Very small images: {len(small_imgs)}")
print(f" Very large images: {len(large_imgs)}")
```

```python
import tensorflow as tf
import numpy as np
from PIL import Image
import os
import glob

# Step 1: Load image paths and labels
dataset_dir = "Tree_Species_Dataset"

image_paths = []
labels = []

for class_name in os.listdir(dataset_dir):
    class_dir = os.path.join(dataset_dir, class_name)
    if os.path.isdir(class_dir):
        for img_file in glob.glob(os.path.join(class_dir, "*.jpg")):
            image_paths.append(img_file)
            labels.append(class_name)

# Step 2: Shuffle dataset
assert len(image_paths) == len(labels)

seed = 42
np.random.seed(seed)

indices = np.arange(len(image_paths))
np.random.shuffle(indices)

image_paths = np.array(image_paths)[indices]
labels = np.array(labels)[indices]

# Step 3: Split into Train / Validation / Test (80 / 10 / 10)
total_size = len(image_paths)

train_size = int(0.8 * total_size)
val_size = int(0.1 * total_size)
test_size = total_size - train_size - val_size

train_paths = image_paths[:train_size]
train_labels = labels[:train_size]

val_paths = image_paths[train_size:train_size + val_size]
val_labels = labels[train_size:train_size + val_size]

test_paths = image_paths[train_size + val_size:]
test_labels = labels[train_size + val_size:]

# Step 4: Convert class names to integer labels
class_names = sorted(set(labels))
label_to_index = {name: idx for idx, name in enumerate(class_names)}

train_labels = np.array([label_to_index[x] for x in train_labels], dtype=np.int32)
val_labels = np.array([label_to_index[x] for x in val_labels], dtype=np.int32)
test_labels = np.array([label_to_index[x] for x in test_labels], dtype=np.int32)

num_classes = len(class_names)

# Step 5: Settings
IMAGE_SIZE = 224
BATCH_SIZE = 32

# Step 6: Load images
# IMPORTANT:
# We DO NOT divide by 255 here.
# Each model will use its own preprocessing later.
def load_and_preprocess_image(path, label):
    image = tf.io.read_file(path)
    image = tf.image.decode_jpeg(image, channels=3)
    image = tf.image.resize(image, (IMAGE_SIZE, IMAGE_SIZE))
    image = tf.cast(image, tf.float32)
    return image, label

# Step 7: Create tf.data datasets
def prepare_dataset(paths, labels, training=False):
    ds = tf.data.Dataset.from_tensor_slices((paths, labels))

    ds = ds.map(
        load_and_preprocess_image,
        num_parallel_calls=tf.data.AUTOTUNE
    )

    if training:
        ds = ds.shuffle(1000, seed=seed)

    ds = ds.batch(BATCH_SIZE)
    ds = ds.cache()
    ds = ds.prefetch(tf.data.AUTOTUNE)

    return ds

train_ds = prepare_dataset(train_paths, train_labels, training=True)
val_ds = prepare_dataset(val_paths, val_labels)
test_ds = prepare_dataset(test_paths, test_labels)

# Step 8: Dataset Summary
print("=" * 50)
print("Dataset Summary")
print("=" * 50)
print(f"Total Images      : {total_size}")
print(f"Training Images   : {len(train_paths)}")
print(f"Validation Images : {len(val_paths)}")
print(f"Testing Images    : {len(test_paths)}")
print(f"Number of Classes : {num_classes}")
print(f"Batch Size        : {BATCH_SIZE}")
print("=" * 50)
```

```python
import tensorflow as tf
from tensorflow.keras import layers

data_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.15),
    layers.RandomZoom(0.15),
    layers.RandomContrast(0.2),
    layers.RandomBrightness(0.15)
])
```

```python
AUTOTUNE = tf.data.AUTOTUNE

# Apply augmentation only to training data.
# Images are already resized in load_and_preprocess_image().

train_ds = train_ds.map(
    lambda x, y: (data_augmentation(x, training=True), y),
    num_parallel_calls=AUTOTUNE
)
```

```python
from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau,
    ModelCheckpoint
)

import time

early_stop = EarlyStopping(
    monitor="val_accuracy",
    mode="max",
    patience=5,
    restore_best_weights=True,
    verbose=1
)

reduce_lr = ReduceLROnPlateau(
    monitor="val_accuracy",
    mode="max",
    factor=0.2,
    patience=2,
    min_lr=1e-6,
    verbose=1
)

checkpoint = ModelCheckpoint(
    "best_model.keras",
    monitor="val_accuracy",
    mode="max",
    save_best_only=True,
    verbose=1
)
```

```python
from tensorflow.keras.applications import MobileNetV2

base_model = MobileNetV2(
    input_shape=(IMAGE_SIZE, IMAGE_SIZE, 3),
    include_top=False,
    weights="imagenet"
)
```

```python
num_classes = num_classes

import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input

# Freeze the backbone initially
base_model.trainable = False

model = models.Sequential([
    # MobileNetV2 expects images in [-1, 1]
    layers.Lambda(preprocess_input),

    base_model,

    layers.GlobalAveragePooling2D(),

    layers.BatchNormalization(),

    layers.Dropout(0.4),

    layers.Dense(
        256,
        activation="relu",
        kernel_regularizer=tf.keras.regularizers.l2(0.001)
    ),

    layers.Dropout(0.3),

    layers.Dense(
        num_classes,
        activation="softmax"
    )
])

model.summary()
```

```python
start_time = time.time()

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=30,
    callbacks=[
        early_stop,
        reduce_lr,
        checkpoint
    ],
    verbose=1
)

training_time = time.time() - start_time

print(f"\nTraining completed in {training_time:.2f} seconds")

# Load the best model weights before evaluation
model.load_weights("best_model.keras")
```

```python
import os

def save_model_versioned(model, folder_path='../Models', base_name='model', extension='.keras'):
    # Ensure the folder exists
    os.makedirs(folder_path, exist_ok=True)

    # Get the list of existing model files in the specified folder
    existing_models = [f for f in os.listdir(folder_path) if f.startswith(base_name) and f.endswith(extension)]

    # Determine the next model version number
    if existing_models:
        # Extract the version number from existing filenames
        versions = [int(f[len(base_name):-len(extension)]) for f in existing_models if f[len(base_name):-len(extension)].isdigit()]
        next_version = max(versions) + 1
    else:
        next_version = 1  # Start with version 1 if no models exist

    # Create the new filename
    new_model_name = os.path.join(folder_path, f"{base_name}{next_version}{extension}")

    # Save the model
    model.save(new_model_name)
    print(f"Model saved as: {new_model_name}")

# Example usage before training
# Assuming you have already trained your model:
# model.fit(...)  # Your training code goes here

# Save the model with versioning after training
save_model_versioned(model, folder_path='../Models', base_name='Model', extension='.keras')
```

```python
from tensorflow.keras.applications import EfficientNetB0

base_model = EfficientNetB0(
    include_top=False,
    weights="imagenet",
    input_shape=(IMAGE_SIZE, IMAGE_SIZE, 3)
)

# Enable fine-tuning
base_model.trainable = True

# Freeze the first 80% of layers
fine_tune_at = int(len(base_model.layers) * 0.8)

for layer in base_model.layers[:fine_tune_at]:
    layer.trainable = False

print("Total layers:", len(base_model.layers))
print("Trainable layers:", sum(layer.trainable for layer in base_model.layers))
print("Frozen layers:", sum(not layer.trainable for layer in base_model.layers))
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.applications.efficientnet import preprocess_input

model = models.Sequential([
    # EfficientNet preprocessing
    layers.Lambda(preprocess_input),

    base_model,

    layers.GlobalAveragePooling2D(),

    layers.Dense(128, activation="relu"),
    layers.Dropout(0.3),

    layers.Dense(64, activation="relu"),
    layers.Dropout(0.2),

    layers.Dense(num_classes, activation="softmax")
])

model.summary()
```

```python
import os

def save_model_versioned(model, model_name, folder_path="../Models"):

    # Create folder if it doesn't exist
    os.makedirs(folder_path, exist_ok=True)

    existing_versions = []

    for file in os.listdir(folder_path):

        if file.startswith(model_name + "_v") and file.endswith(".keras"):

            try:
                version = int(
                    file.replace(model_name + "_v", "")
                        .replace(".keras", "")
                )
                existing_versions.append(version)

            except ValueError:
                pass

    next_version = max(existing_versions, default=0) + 1

    filename = f"{model_name}_v{next_version}.keras"

    save_path = os.path.join(folder_path, filename)

    model.save(save_path)

    print("=" * 60)
    print("✅ Model Saved Successfully")
    print(f"Model Name : {filename}")
    print(f"Location   : {os.path.abspath(save_path)}")
    print("=" * 60)

    return save_path
```
